```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/cleaned_jobs.csv")
```

```python
df["work_year"].value_counts().sort_index()
```

```python
year_data = df["work_year"].value_counts().sort_index()

plt.figure(figsize=(8,5))

year_data.plot(kind="line", marker="o")

plt.title("Jobs Posted by Year")
plt.xlabel("Year")
plt.ylabel("Count")

plt.show()
```

```python
df.groupby("work_year")["salary_in_usd"].mean()
```

```python
salary_year = df.groupby("work_year")["salary_in_usd"].mean()

plt.figure(figsize=(8,5))

salary_year.plot(kind="line", marker="o")

plt.title("Average Salary by Year")
plt.ylabel("Salary USD")

plt.show()
```

```python
df["company_size"].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="company_size",
    data=df
)

plt.title("Company Size Distribution")

plt.show()
```

```python
df["remote_ratio"].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="remote_ratio",
    data=df
)

plt.title("Remote Work Distribution")

plt.show()
```
