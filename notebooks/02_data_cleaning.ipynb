```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/jobs.csv")
```

```python
print("Before Cleaning:")
print(df.shape)
```

```python
df.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
df.columns
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
experience_map = {
    "EN":"Entry Level",
    "MI":"Mid Level",
    "SE":"Senior Level",
    "EX":"Executive Level"
}

df["experience_level_name"] = df["experience_level"].map(experience_map)
```

```python
df[["experience_level","experience_level_name"]].head()
```

```python
employment_map = {
    "FT":"Full Time",
    "PT":"Part Time",
    "CT":"Contract",
    "FL":"Freelance"
}

df["employment_type_name"] = df["employment_type"].map(employment_map)
```

```python
df[["employment_type","employment_type_name"]].head()
```

```python
df["salary_category"] = pd.cut(
    df["salary_in_usd"],
    bins=[0,50000,100000,150000,500000],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["salary_category"].value_counts()
```

```python
df.head()
```

```python
df.to_csv(
    "../data/processed/cleaned_jobs.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
