```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/cleaned_jobs.csv")

df.head()
```

```python
df["experience_level_name"].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="experience_level_name",
    data=df
)

plt.title("Experience Level Distribution")
plt.xticks(rotation=15)

plt.show()
```

```python
plt.savefig(
    "../visualizations/exp_level.png",
    bbox_inches="tight"
)
```

```python
plt.savefig(
    "../visualizations/emp_type.png",
    bbox_inches="tight"
)
```

```python
df["job_title"].value_counts().head(10)
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    y="job_title",
    data=df,
    order=df["job_title"].value_counts().head(10).index
)

plt.title("Top 10 Job Titles")

plt.show()
```

```python
plt.savefig(
    "../visualizations/top_j10_ob_titles.png",
    bbox_inches="tight"
)
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["salary_in_usd"],
    bins=30,
    kde=True
)

plt.title("Salary Distribution")

plt.show()
```

```python
plt.savefig(
    "../visualizations/salary_distribution.png",
    bbox_inches="tight"
)
```

```python
df.groupby(
    "experience_level_name"
)["salary_in_usd"].mean().sort_values(ascending=False)
```

```python
avg_salary = df.groupby(
    "experience_level_name"
)["salary_in_usd"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,5))

avg_salary.plot(kind="bar")

plt.title("Average Salary by Experience Level")
plt.ylabel("Salary in USD")

plt.show()
```

```python
plt.savefig(
    "../visualizations/avg_salary_experience.png",
    bbox_inches="tight"
)
```
